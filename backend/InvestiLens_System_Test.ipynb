## Section 10 — Full Pipeline Simulation

Runs all analysis chains in sequence, like the orchestrator does.  
Set `ANALYSIS_COMPANY` to a company you have already ingested in Section 7 for real results.

```python
import json
from app.rag.retriever import retrieve_for_analysis
from app.rag.context_builder import build_context
from app.llm import chains

# ---- Set to a company you have ingested, or keep MOCK_COMPANY for demo ------
ANALYSIS_COMPANY = MOCK_COMPANY
# -----------------------------------------------------------------------------

print(f"Running pipeline simulation for: {ANALYSIS_COMPANY}")

# Use real retrieval if documents exist, otherwise fall back to mock context
try:
    broad_queries = [
        f"{ANALYSIS_COMPANY} business model, products, and overview",
        f"{ANALYSIS_COMPANY} financial performance, revenue, profit, cash flow",
        f"{ANALYSIS_COMPANY} strategic priorities, investments, R&D",
        f"{ANALYSIS_COMPANY} risks, threats, competition, regulation",
        f"{ANALYSIS_COMPANY} management outlook, future expectations"
    ]
    chunks = retrieve_for_analysis(queries=broad_queries, company=ANALYSIS_COMPANY, years=None, top_k_per_query=3)
    if chunks:
        context = build_context(chunks)
        print(f"Retrieved {len(chunks)} chunks from ChromaDB")
    else:
        context = MOCK_CONTEXT
        print("No chunks found in ChromaDB — using mock context")
except Exception as e:
    context = MOCK_CONTEXT
    print(f"Retrieval error ({e}) — using mock context")

print()

pipeline_results = {}
steps = [
    ("Company Analysis",    lambda: chains.run_company_analysis(ANALYSIS_COMPANY, context)),
    ("Financial Analysis",  lambda: chains.run_financial_analysis(ANALYSIS_COMPANY, context)),
    ("Risk Analysis",       lambda: chains.run_risk_analysis(ANALYSIS_COMPANY, context)),
    ("Strategy Analysis",   lambda: chains.run_strategy_analysis(ANALYSIS_COMPANY, context)),
    ("Management Outlook",  lambda: chains.run_management_outlook(ANALYSIS_COMPANY, context)),
]

for name, fn in steps:
    print(f"  Running {name}...", end=" ", flush=True)
    try:
        result = fn()
        pipeline_results[name] = result
        print("DONE")
    except Exception as e:
        print(f"FAILED: {e}")
        pipeline_results[name] = None

print()
print("=" * 55)
print("  Pipeline Summary")
print("=" * 55)
for name, result in pipeline_results.items():
    status = "PASS" if result is not None else "FAIL"
    print(f"  [{status}] {name}")
print()
print("Results stored in: pipeline_results")
```

```python
import httpx
import os
import json
import time

# ============================================================
# CONFIGURATION
# ============================================================

BASE_URL = "http://127.0.0.1:8000"

# Change this to your actual PDF
PDF_PATH = r"C:\Users\bisoi\Downloads\InvestiLens\backend\data\documents\Apple_2024_annual_report.pdf"

# Test company
COMPANY_DATA = {
    "name": "Apple Inc.",
    "industry": "Technology",
    "ticker": "AAPL",
    "country": "United States",
    "analysis_period": "2024"
}

# Test question for RAG
CHAT_QUESTION = (
    "What were Apple's main financial highlights and risks "
    "mentioned in the annual report?"
)


# ============================================================
# HELPERS
# ============================================================

def print_result(step, success, status_code=None, data=None, error=None):
    symbol = "✓ PASS" if success else "✗ FAIL"

    if status_code is not None:
        print(f"{symbol:<10} {step:<30} [{status_code}]")
    else:
        print(f"{symbol:<10} {step}")

    if data is not None:
        print("           ", str(data)[:250])

    if error:
        print("           ERROR:", error)


def get_json(response):
    try:
        return response.json()
    except Exception:
        return response.text


# ============================================================
# START TEST
# ============================================================

print("=" * 80)
print("              INVESTILENS API END-TO-END TEST")
print("=" * 80)

company_id = None
document_id = None
analysis_id = None


with httpx.Client(
    base_url=BASE_URL,
    timeout=120
) as client:

    # ========================================================
    # 1. HEALTH CHECK
    # ========================================================

    try:
        response = client.get("/health")
        data = get_json(response)

        print_result(
            "Health Check",
            response.status_code == 200,
            response.status_code,
            data
        )

    except Exception as e:
        print_result(
            "Health Check",
            False,
            error=str(e)
        )


    # ========================================================
    # 2. CREATE COMPANY
    # ========================================================

    try:
        response = client.post(
            "/api/companies/",
            json=COMPANY_DATA
        )

        data = get_json(response)

        success = response.status_code < 400

        if success and isinstance(data, dict):

            # Try common possible response structures
            company_id = (
                data.get("id")
                or data.get("company_id")
            )

            if company_id is None and isinstance(data.get("data"), dict):
                company_id = (
                    data["data"].get("id")
                    or data["data"].get("company_id")
                )

        print_result(
            "Create Company",
            success,
            response.status_code,
            data
        )

        print("           Company ID:", company_id)

    except Exception as e:
        print_result(
            "Create Company",
            False,
            error=str(e)
        )


    # ========================================================
    # 3. GET COMPANY
    # ========================================================

    if company_id:

        try:
            response = client.get(
                f"/api/companies/{company_id}"
            )

            data = get_json(response)

            print_result(
                "Get Company",
                response.status_code == 200,
                response.status_code,
                data
            )

        except Exception as e:
            print_result(
                "Get Company",
                False,
                error=str(e)
            )

    else:
        print_result(
            "Get Company",
            False,
            error="company_id was not returned from Create Company"
        )


    # ========================================================
    # 4. UPLOAD DOCUMENT
    # ========================================================

    if company_id and os.path.exists(PDF_PATH):

        try:

            with open(PDF_PATH, "rb") as pdf_file:

                files = {
                    "file": (
                        os.path.basename(PDF_PATH),
                        pdf_file,
                        "application/pdf"
                    )
                }

                form_data = {
                    "company_id": str(company_id),
                    "document_name": "Apple 2024 Annual Report",
                    "document_type": "annual_report",
                    "year": "2024"
                }

                response = client.post(
                    "/api/documents/upload",
                    data=form_data,
                    files=files
                )

            data = get_json(response)

            success = response.status_code < 400

            if success and isinstance(data, dict):

                document_id = (
                    data.get("id")
                    or data.get("document_id")
                )

                if document_id is None and isinstance(data.get("data"), dict):
                    document_id = (
                        data["data"].get("id")
                        or data["data"].get("document_id")
                    )

            print_result(
                "Upload Document",
                success,
                response.status_code,
                data
            )

            print("           Document ID:", document_id)

        except Exception as e:
            print_result(
                "Upload Document",
                False,
                error=str(e)
            )

    elif not os.path.exists(PDF_PATH):

        print_result(
            "Upload Document",
            False,
            error=f"PDF not found: {PDF_PATH}"
        )

    else:

        print_result(
            "Upload Document",
            False,
            error="company_id is missing"
        )


    # ========================================================
    # 5. LIST COMPANY DOCUMENTS
    # ========================================================

    if company_id:

        try:
            response = client.get(
                f"/api/documents/company/{company_id}"
            )

            data = get_json(response)

            print_result(
                "List Company Documents",
                response.status_code == 200,
                response.status_code,
                data
            )

        except Exception as e:
            print_result(
                "List Company Documents",
                False,
                error=str(e)
            )


    # ========================================================
    # 6. START ANALYSIS
    # ========================================================

    if company_id:

        try:

            response = client.post(
                "/api/analysis/start",
                params={
                    "company_id": company_id,
                    "years": "2024"
                }
            )

            data = get_json(response)

            success = response.status_code < 400

            if success and isinstance(data, dict):

                analysis_id = (
                    data.get("id")
                    or data.get("analysis_id")
                )

                if analysis_id is None and isinstance(data.get("data"), dict):
                    analysis_id = (
                        data["data"].get("id")
                        or data["data"].get("analysis_id")
                    )

            print_result(
                "Start Analysis",
                success,
                response.status_code,
                data
            )

            print("           Analysis ID:", analysis_id)

        except Exception as e:
            print_result(
                "Start Analysis",
                False,
                error=str(e)
            )


    # ========================================================
    # 7. GET ANALYSIS
    # ========================================================

    if analysis_id:

        try:

            response = client.get(
                f"/api/analysis/{analysis_id}"
            )

            data = get_json(response)

            print_result(
                "Get Analysis",
                response.status_code == 200,
                response.status_code,
                data
            )

        except Exception as e:
            print_result(
                "Get Analysis",
                False,
                error=str(e)
            )


    # ========================================================
    # 8. GET COMPANY ANALYSES
    # ========================================================

    if company_id:

        try:

            response = client.get(
                f"/api/analysis/company/{company_id}"
            )

            data = get_json(response)

            print_result(
                "Get Company Analyses",
                response.status_code == 200,
                response.status_code,
                data
            )

        except Exception as e:
            print_result(
                "Get Company Analyses",
                False,
                error=str(e)
            )


    # ========================================================
    # 9. RAG CHAT
    # ========================================================

    if company_id:

        try:

            chat_payload = {
                "company_id": str(company_id),
                "question": CHAT_QUESTION,
                "filters": None
            }

            response = client.post(
                "/api/chat/",
                json=chat_payload
            )

            data = get_json(response)

            print_result(
                "RAG Chat",
                response.status_code < 400,
                response.status_code,
                data
            )

        except Exception as e:
            print_result(
                "RAG Chat",
                False,
                error=str(e)
            )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("                    TEST SUMMARY")
print("=" * 80)

print(f"Company ID  : {company_id}")
print(f"Document ID : {document_id}")
print(f"Analysis ID : {analysis_id}")

print("=" * 80)
print("Swagger UI:")
print(f"{BASE_URL}/docs")
print("=" * 80)
```
